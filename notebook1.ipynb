```python
# Задача1: Принимать три параметра (a: int, b: int, length: int) и возвращать список (list) длины length из равномерно
# распределенной случайной величины между числами a и b

# функция
def random_list(a, b, length):
    return list(random.randint(a, b) for _ in range(length))
    # return [random.randint(a, b) for _ in range(length)]

# юзерский ввод, проверка на дурака и замена на [1,10,5] если некорректно
a = input("Введите начало A: ")
b = input("Введите конец B: ")
length = input("Введите длину L: ")
comment = ""
if not b.isdigit():
    b = 10
else:
    b = int(b)    
if not a.isdigit() or int(a) > int(b):
    a = 1
else:
    a = int(a)    
if not length.isdigit() or int(length) < 1:
    length = 5
else:
    length = int(length)     
print("A = ", a)    
print("B = ", b)       
print("L = ", length)

# вывод результата 1
result_random_list = random_list(a, b, length)
print("Рандомный список:", result_random_list, sep='\n')

# Задача2: Принимать созданный список (list) и считать количество уникальных значений переменной (строить частотный
# ряд) в виде dict. Например, для списка [1, 2, 3, 1, 2, 2] будет возвращать {1 : 2, 2 : 3, 3 : 1}

# функция
def frequency_dict(result_random_list):
    freq = {}
    # for i in range(len(result_random_list)):
    #     item = result_random_list[i]
    for item in result_random_list:
        freq[item] = freq.get(item, 0) + 1
    return freq

# вывод результата 2
result_frequency_dict = frequency_dict(result_random_list)
# print("Частотный словарь = ", result_frequency_dict)
print("Частотный словарь:", result_frequency_dict, sep='\n')
# print("Частотный словарь = /n", result_frequency_dict)
# print("Частотный словарь = ")
# print(result_frequency_dict)
# print("Частотный словарь =", result_frequency_dict, sep='\n')
# print(f"Частотный словарь =\n{result_frequency_dict}")

# Задача3: Считать минимум, среднюю, медиану, моду, максимум для значений списка

# функция
import statistics # библиотека для расчетов
def list_statistics(result_random_list):
    list_min = min(result_random_list)
    list_max = max(result_random_list)
    # list_ave = sum(result_random_list) / len(result_random_list)
    # list_ave = int(list_ave) # округляем до целого 
    # list_ave = round(list_ave) # округляем до целого 
    list_ave = int(statistics.mean(result_random_list))
    list_med = int(statistics.median(result_random_list))
    list_mod = int(statistics.mode(result_random_list))
    return {
        'Минимум': list_min,
        'Максимум': list_max,
        'Среднее': list_ave,
        'Медиана': list_med,
        'Мода': list_mod
    }

# вывод результата 3
result_list_statistics = list_statistics(result_random_list)
print("Статистика списка:")
for key in result_list_statistics:
    print(f"{key} = {result_list_statistics[key]}")
```
